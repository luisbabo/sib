```python
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.logistic_regression import LogisticRegression

caminho_breast = "/home/luisbabo/Documentos/Bioinformatica_2ano/sib/datasets/breast_bin/breast-bin.csv" 
dataset_breast = read_csv(caminho_breast, sep=",", features=True, label=True)

train_dataset, test_dataset = train_test_split(dataset_breast, test_size=0.2, random_state=42)

log_reg = LogisticRegression(l2_penalty=1.0, alpha=0.001, max_iter=2000, patience=5, scale=True)

log_reg.fit(train_dataset)

score_final = log_reg.score(test_dataset)
custo_final = log_reg.cost(test_dataset)

print("--- Resultados do Teste LogisticRegression ---")
print(f"Score (Accuracy): {score_final}")
print(f"Cost (Log Loss): {custo_final}")
```
