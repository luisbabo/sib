```python
# imports
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split, stratified_train_test_split
from si.models.knn_classifier import KNNClassifier
```

```python
# datasets
iris_dataset = read_csv('../datasets/iris/iris.csv', features=True, label=True)
```

```python
import numpy as np

# 1. Aplicar a função de divisão estratificada
train_dataset_strat, test_dataset_strat = stratified_train_test_split(iris_dataset, test_size=0.2, random_state=42)

# 2. Contar as ocorrências de cada classe no dataset original, de treino e de teste
_, contagem_original = np.unique(iris_dataset.y, return_counts=True)
_, contagem_treino = np.unique(train_dataset_strat.y, return_counts=True)
_, contagem_teste = np.unique(test_dataset_strat.y, return_counts=True)

# 3. Imprimir os resultados para validar a proporção
print(f"Distribuição Original: {contagem_original}")
print(f"Distribuição no Treino: {contagem_treino}")
print(f"Distribuição no Teste: {contagem_teste}")
```

```python
# split dataset
train_dataset, test_dataset = train_test_split(iris_dataset)
test_dataset.X
```

```python
# knn classifier
k = 2
kmeans = KNNClassifier(k)
kmeans.fit(train_dataset)
predictions = kmeans.predict(test_dataset)
predictions
```

```python
kmeans.score(test_dataset)
```
